## Here are two valid and complementary ways to use SHAP:

• Training set SHAP: explains what patterns and feature relationships the model learned during training (interpretability of the fitted model).
• Test set SHAP: validates that feature importance and directionality are consistent on unseen data, supporting generalization and robustness.

Imports & setup

```python
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt
from joblib import load
from sklearn.model_selection import train_test_split
```

Load dataset & define target/weight

```python
df = pd.read_pickle("../data/prepared/NSDUH_2023_prepared.pkl")

TARGET = "IRAMDEYR"
WEIGHT = "ANALWT2_C"

df[TARGET] = (df[TARGET] == 1).astype(int)

y = df[TARGET]
w = df[WEIGHT]
```

Load Tier 2 / Tier 3 feature lists

```python
from tiering_preparation import TIERS
```

Define tiers & model paths

```python
tiers = {
    "Tier_2_Clinical": {
        "features": TIERS["Tier_2_Clinical"],
        "model_path": "../models/lightgbm_tier_2_clinical_recall.joblib",
    },
    "Tier_3_Personalized": {
        "features": TIERS["Tier_3_Personalized"],
        "model_path": "../models/lightgbm_tier_3_personalized_recall.joblib",
    },
}
```

Loop & run SHAP

```python
for tier_name, cfg in tiers.items():

    print(f"\n===== Running SHAP for {tier_name} =====")

    features = cfg["features"]
    model_path = cfg["model_path"]

    missing_features = set(features) - set(df.columns)
    if missing_features:
        raise ValueError(
            f"{tier_name} references missing columns: {missing_features}"
        )

    X = df[features]

    print(f"\n{tier_name} — Missing values per feature:")
    print(X.isnull().sum().sort_values(ascending=False))



    X_train, X_test, y_train, y_test, w_train, w_test = train_test_split(
        X, y, w,
        test_size=0.2,
        stratify=y,
        random_state=42
    )

    model = load(model_path)
    explainer = shap.TreeExplainer(model)

    datasets = {
        "Train": (X_train, y_train),
        "Test": (X_test, y_test),
    }

    for split_name, (X_split, y_split) in datasets.items():

        print(f"{tier_name} | {split_name} | X shape: {X_split.shape}")

        shap_values = explainer.shap_values(X_split)

        if isinstance(shap_values, list):
            shap_values_pos = shap_values[1]
            base_value = explainer.expected_value[1]
        else:
            shap_values_pos = shap_values
            base_value = explainer.expected_value

        print(
            f"{tier_name} | {split_name} | SHAP shape: {shap_values_pos.shape}"
        )

        # SHAP Explanation object (modern API)
        shap_exp = shap.Explanation(
            values=shap_values_pos,
            base_values=base_value,
            data=X_split,
            feature_names=X_split.columns
        )

        # Beeswarm (global + direction)
        shap.summary_plot(
            shap_exp.values,
            X_split,
            show=False
        )
        plt.title(f"SHAP Summary — {tier_name} ({split_name})")
        plt.tight_layout()
        plt.show()

        # Bar plot (global importance)
        shap.summary_plot(
            shap_exp.values,
            X_split,
            plot_type="bar",
            show=False
        )
        plt.title(f"SHAP Feature Importance — {tier_name} ({split_name})")
        plt.tight_layout()
        plt.show()

        # Partial Dependence / PDP-style plot (top feature)
        # Number of top features to plot
        TOP_K = 20

        # Compute mean absolute SHAP per feature
        mean_abs_shap = np.abs(shap_values_pos).mean(axis=0)

        # Get indices of top K features
        top_k_idx = np.argsort(mean_abs_shap)[-TOP_K:][::-1]

        # Create SHAP Explanation object (important)
        shap_exp = shap.Explanation(
            values=shap_values_pos,
            base_values=base_value,
            data=X_split.values,
            feature_names=X_split.columns
        )

        # Loop through top K features
        for idx in top_k_idx:
            feature_name = X_split.columns[idx]

            print(
                f"{tier_name} | {split_name} | Dependence plot for: {feature_name}"
            )

            shap.plots.scatter(
                shap_exp[:, idx],
                show=True
            )
```
